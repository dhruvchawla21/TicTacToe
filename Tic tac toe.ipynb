```python
matrix = [
    ["0","|","1","|","2"],
    ["-","+","-","+","-"],
    ["3","|","4","|","5"],
    ["-","+","-","+","-"],
    ["6","|","7","|","8"]
]
def mat():
    for row in matrix:
        for elem in row:
            print(elem, end=' ')
        print() 
mat()
```

```python
marker = input("Choose ur marker X or O :")
if marker == "X":
    S = "X"
    Q = "O"
else :
    S = "O"
    Q = "X"
```

```python
def player1():
    x=int(input("Take ur turn Player 1 : "))
    for i in range(0,5,2):
        for j in range(0,5,2):
            if matrix[i][j]== str(x):
                matrix[i][j] = S
    mat()


    
```

```python
def player2():
    x=int(input("Take ur turn Player 2 : "))
    for i in range(0,5,2) :
        for j in range(0,5,2):
            if matrix[i][j]== str(x):
                matrix[i][j] = Q
                
                
    mat()
```

```python
def rules():
    # Rows
    for i in range(0,5,2):
        if matrix[i][0] == matrix[i][2] == matrix[i][4]:
            return matrix[i][0]

    # Columns
    for i in range(0,5,2):
        if matrix[0][i] == matrix[2][i] == matrix[4][i]:
            return matrix[0][i]

    # Diagonals
    if matrix[0][0] == matrix[2][2] == matrix[4][4]:
        return matrix[0][0]

    if matrix[0][4] == matrix[2][2] == matrix[4][0]:
        return matrix[2][2]

    return None
        

```

```python
mat()
for turn in range(9):
    if turn % 2 == 0:
        player1()
    else:
        player2()

    winner = rules()
    if winner:
        print(winner, "wins!")
        break
if winner == None : 
    print("DRAW")
```
